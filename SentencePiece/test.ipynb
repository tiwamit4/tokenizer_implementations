```python
# pip install sentencepiece

import sentencepiece as spm
from pathlib import Path

# Train a BPE SentencePiece model from a plain-text training corpus.
corpus_file = Path("data/train.txt")
model_prefix = "models/sentencepiece_bpe"

spm.SentencePieceTrainer.train(
    input=str(corpus_file),
    model_prefix=str(model_prefix),
    vocab_size=8_000,
    model_type="bpe",          # Alternatives: "unigram", "char", "word"
    character_coverage=0.9995, # Useful for multilingual text
    unk_id=0,
    bos_id=1,
    eos_id=2,
    pad_id=3,
)

# Load and use the trained tokenizer.
tokenizer = spm.SentencePieceProcessor(model_file=f"{model_prefix}.model")

text = "SentencePiece tokenizes text without requiring whitespace splitting."

token_ids = tokenizer.encode(text, out_type=int)
pieces = tokenizer.encode(text, out_type=str)
decoded_text = tokenizer.decode(token_ids)

print("Token IDs:", token_ids)
print("Pieces:", pieces)
print("Decoded:", decoded_text)

# Batch encoding
texts = ["Hello world!", "Tokenizers turn text into token IDs."]
batch_ids = [tokenizer.encode(item, out_type=int) for item in texts]

# Special-token-aware encoding
with_bos_eos = [tokenizer.bos_id()] + token_ids + [tokenizer.eos_id()]
print("With BOS/EOS:", with_bos_eos)
```
